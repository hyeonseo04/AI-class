```python
# ============================================================================
# Wine - Regression (MSE)
# 모델: Linear Regression, Decision Tree, Random Forest, KNN
# 타겟: alcohol (1번째 컬럼), Feature: 나머지 12개
# ============================================================================

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.datasets import load_wine
```

```python
# ============================================================================
# 1. 데이터 불러오는 방식 3가지
# ============================================================================

# --- 방법 1: sklearn 내장 데이터셋 ---
wine = load_wine()
X1 = wine.data[:, 1:]     # alcohol 제외한 12개 feature
y1 = wine.data[:, 0]      # alcohol (연속값)
print("방법 1 (sklearn 내장):", X1.shape, y1.shape)

# --- 방법 2: csv 저장 후 pd.read_csv ---
df = pd.DataFrame(wine.data, columns=wine.feature_names)
df.to_csv('wine_reg.csv', index=False)

wine_csv = pd.read_csv('wine_reg.csv')
X2 = wine_csv.iloc[:, 1:].values
y2 = wine_csv.iloc[:, 0].values
print("방법 2 (csv 저장→읽기):", X2.shape, y2.shape)

# --- 방법 3: URL에서 직접 불러오기 ---
try:
    url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data"
    col_names = ['class'] + list(wine.feature_names)  # UCI wine: 첫 열이 class
    wine_web = pd.read_csv(url, header=None, names=col_names)
    X3 = wine_web.drop(['class', 'alcohol'], axis=1).values
    y3 = wine_web['alcohol'].values
    print("방법 3 (URL 직접):", X3.shape, y3.shape)
except Exception as e:
    print(f"방법 3 (URL 직접): 네트워크 오류 - {e}")
```

```python
# 'alcohol' 열을 타겟으로 설정하고 나머지를 특성으로 설정
y = df['alcohol']
X = df.drop('alcohol', axis=1)
```

```python
# 학습 / 테스트 분리 (75% / 25%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
```

```python

# ── 1. Linear Regression ────────────────────────────
model_lr = LinearRegression()
model_lr.fit(X_train, y_train)
pred_lr = model_lr.predict(X_test)

# ── 2. Decision Tree ────────────────────────────────
model_dt = DecisionTreeRegressor(random_state=0)
model_dt.fit(X_train, y_train)
pred_dt = model_dt.predict(X_test)

# ── 3. Random Forest ────────────────────────────────
model_rf = RandomForestRegressor(random_state=0, n_estimators=100)
model_rf.fit(X_train, y_train)
pred_rf = model_rf.predict(X_test)

# ── 4. KNN ──────────────────────────────────────────
model_knn = KNeighborsRegressor(n_neighbors=5)
model_knn.fit(X_train, y_train)
pred_knn = model_knn.predict(X_test)
```

```python

# ── MSE 계산 함수 3가지 ──────────────────────────────
# ① sklearn
def mse_sklearn(actual, predicted):
    return mean_squared_error(actual, predicted)

# ② numpy
def mse_numpy(actual, predicted):
    return np.mean((np.array(actual) - np.array(predicted)) ** 2)

# ③ 순수 Python
def mse_python(actual, predicted):
    return sum((a - p) ** 2 for a, p in zip(actual, predicted)) / len(actual)
```

```python
models = {
    "Linear Regression": pred_lr,
    "Decision Tree":     pred_dt,
    "Random Forest":     pred_rf,
    "KNN":               pred_knn,
}

for name, pred in models.items():
    print(f"\n--- {name} ---")
    print(f"MSE (sklearn) : {mse_sklearn(y_test, pred):.4f}")
    print(f"MSE (numpy)   : {mse_numpy(y_test, pred):.4f}")
    print(f"MSE (python)  : {mse_python(y_test, pred):.4f}")
```
